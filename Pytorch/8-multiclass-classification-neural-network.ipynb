```python
from sklearn.datasets import load_iris
import pandas as pd
import numpy as np
```

```python
iris = load_iris()
```

```python
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['target'] = iris.target
df.head()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

```python
train_df,test_df = train_test_split(df,test_size=0.2,random_state=42,stratify=df['target'])
```

```python
X_train = train_df.drop('target', axis=1)
y_train = train_df['target']

X_test = test_df.drop('target', axis=1)
y_test = test_df['target']
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train[0]
```

```python
X_train_tensor = torch.tensor(X_train,dtype=torch.float32)
X_test_tensor = torch.tensor(X_test,dtype=torch.float32)
y_train_tensor = torch.tensor(y_train.values,dtype=torch.long)
y_test_tensor = torch.tensor(y_test.values,dtype=torch.long)
```

```python
class IrisClassifier(nn.Module):
    
    def __init__(self,input_size,hidden_size,output_size):
        super(IrisClassifier,self).__init__()

        self.network = nn.Sequential(
            nn.Linear(input_size,hidden_size),
            nn.ReLU(),
            nn.Linear(hidden_size,hidden_size),
            nn.ReLU(),
            nn.Linear(hidden_size,output_size),
        )

    def forward(self,x):
        return self.network(x)
```

```python
input_size = X_train_tensor.shape[1]
hidden_size = 16
output_size = 3

```

```python
model = IrisClassifier(input_size,hidden_size,output_size)
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(),lr=0.01)
```

```python
## Model Training
epoch = 500
for e in range(epoch):
    model.train()
    optimizer.zero_grad()
    
    predictions = model(X_train_tensor)
    loss=criterion(predictions,y_train_tensor)

    loss.backward()
    optimizer.step()

    if(e+1)%10==0:
        print(f"Epoch [{e+1}] / {epoch} , loss : {loss.item():.4f}")
```

```python
model.eval()
with torch.no_grad():
    y_pred = model(X_test_tensor)
    y_pred_labels = torch.argmax(y_pred,dim=1) ## Class with highest prob

    accuracy = (y_pred_labels == y_test_tensor).sum().item() / y_test_tensor.size(0)

    print(f"Test Acuuracy is : {accuracy}")
```

```python
def predict_iris(sepal_length,sepal_width,petal_length,petal_width):
    input_data = np.array([[sepal_length,sepal_width,petal_length,petal_width]])
    input_data = scaler.transform(input_data)
    input_data = torch.tensor(input_data, dtype=torch.float32)

    model.eval()
    with torch.no_grad():
        prediction = model(input_data)
        prediction_label = torch.argmax(prediction,dim=1).item()
    return prediction_label
```

```python
predict_iris(5.1,3.5,1.4,0.2)
```
