```python
import torch
```

```python
tensor = torch.arange(12)
```

```python
reshaped_tensor_view = tensor.view(2,6)
reshaped_tensor_reshape = tensor.reshape(2,6)
```

```python
print(reshaped_tensor_view)
print(reshaped_tensor_reshape)
```

```python
tensor.is_contiguous()
```

```python
tensor.view(4,-1)
```

```python
tensor_3d = torch.arange(24).reshape(2,3,4)
print(tensor_3d)
```

```python
flattened = tensor_3d.reshape(-1)
print(flattened)
```

```python
flattened.is_contiguous()
```

```python
tensor_2d = torch.arange(24).reshape(12,2)
transpose_tensor = tensor_2d.t()
print(transpose_tensor)
```

```python
transpose_tensor.is_contiguous()
```

```python
error = transpose_tensor.view(6,4)
```
