# Splitting and Joining

```python
'''
torch.cat() --> merges tensors along an existing dimension
'''

ten1 = torch.tensor([[1,2],[3,4]])
ten2 = torch.tensor([[5,6],[7,8]])

torch.cat((ten1,ten2),dim=0)
```

```python
torch.cat((ten1,ten2),dim=1)
```

```python
# Stack
#Creates a new Dimension which increases the tensor rank
print(torch.stack((ten1,ten2),dim=0)) ## Stacking along rows
print("------"*20)
print(torch.stack((ten1,ten2),dim=1))  ## Stacking along columns
```

```python
'''
torch.chunk() -> Divides tensor into equal size chunks
torch.split() -> Allows uneven splitting based on size
'''

ten3 = torch.arange(12)
print(ten3)
```

```python
chunks=torch.chunk(ten3,3,dim=0)
print(chunks) ## tuple

for chunk in chunks:
    print(chunk)
```

```python
splits=torch.split(ten3,5,dim=0)
print(splits) ## tuple

for split in splits:
     print(split)
```

# Transposing and Permuting
- transpose() --> swaps 2 dimensio. mxn --> nxm

- permute( --> rearranges all dimension in specified order

```python
ten4 = torch.arange(24).reshape(12,2)
print(ten4)
```

```python
transpose_tensor = ten4.transpose(0,1)
print(transpose_tensor)
```

```python
ten5 = torch.arange(24).reshape(2,3,4)
print(ten5)
```

```python
ten5.permute(2,0,1) ## Sequence 2-4, 0-2, 1-3
```

# Cloning AND Detaching

```python
ten6 = torch.ones(3,3)
```

```python
print(ten6) ## part of computaion graph
```

```python
cloned_tens = ten6.clone() ## part of computaion graph
print(cloned_tens)
```

```python
detached_tensor = ten6.detach() ## this will detach tensor from computaional graph 
print(detached_tensor)
# not a part of computaional graph but storage will be same as originals 
```
