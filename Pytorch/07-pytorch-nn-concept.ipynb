### Simple Neural Network Breakdown

- Input X
- Weights W
- Bias B
- Activation A
- Output Y

  Z = W1.X+B1
  
  Z' = A(Z)

  Y = A(W2.Z'+B2)

LOSS --> Gradients ---> Back Prpogation --> Optimizers

## Components of Pytorch
- Base Class for defining custom Models : torch.nn.Module
- Fully connected (dense layers) : torch.nn.Linear
- Activation Fuction : torch.nn.ReLU
- Optimizers : torch.optim
- Loss Function : torch.nn.CrossEntropyLoss
- Loads Data in Batches : torch.utils.daya.DataLoader
  

## Ways to create a NN
1. Function : Flexible, little harder to interpret
2. Sequential : More structured --> nn.Sequential()

# Building A Neural Network

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
## Functional API

class SimpleNN(nn.Module):
    
    def __init__(self,input_size,hidden_size,output_size):
        super(SimpleNN,self).__init__()

        self.fc1 = nn.Linear(input_size,hidden_size)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_size,output_size)

    def forward(self,x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        return x
        
```

```python
## Sequential API

class SimpleNNSequential(nn.Module):
    
    def __init__(self,input_size,hidden_size,output_size):
        super(SimpleNN,self).__init__()

        self.network = nn.Sequential(
            nn.Linear(input_size,hidden_size),
            nn.ReLU(),
            nn.linear(hidden_size,output_size)
        )
    
    def forward(self,x):
       x = self.network(x)
       return x
        
```

## Training The Neural Network

```python
model_func = SimpleNN(input_size=4,hidden_size=8,output_size=3)
print(model_func)
```

```python
## dummy data
X = torch.randn(10,4) ## 10 sample with 4 features
y = torch.randint(0,3,(10,)) ## class labels

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model_func.parameters(),lr=0.01)
```

```python
print(X)
print("\n")
print(y)
```

```python
## Training LOOP

epoch = 120
for e in range(epoch):
    optimizer.zero_grad()
    outputs = model_func(X)
    loss = criterion(outputs,y)
    loss.backward()
    optimizer.step() ## updation of weights

    if(e+1) % 10 == 0:
        print(f"Epoch[{e+1}]/120 , Loss : {loss.item() :.4f}")
```
