# What are Tensors?
#### PyTorch tensors are a fundamental data structure in PyTorch, designed to handle multi-dimensional arrays. They are similar to NumPy arrays but come with additional features like GPU acceleration, automatic differentiation, and seamless integration with deep learning models.


```python
import torch
```

```python
#1. Scaler Tensor (tensor with 0 dimension)
scaler=torch.tensor(42)
print("Scaler Tensor : ",scaler)
print("Scaler Dimension : ",scaler.ndim)
print("Scaler Tensor Shape : ",scaler.shape)
```

```python
#2. One dimension tensor (Vector)

vector = torch.tensor([1,2,3])
print("vector Tensor : ",vector)
print("vector Dimension : ",vector.ndim)
print("vector Tensor Shape : ",vector.shape)
```

```python
#3. Matrix

matrix = torch.tensor([[1,2,3],[4,5,6]])
print("matrix Tensor : ",matrix)
print("matrix Dimension : ",matrix.ndim)
print("matrix Tensor Shape : ",matrix.shape)
```

```python
#4. 3d Tensor (does not have any specific name)
three_D_tensor = torch.tensor([[[1,2],[3,4],[5,6]]])
print("three_D_tensor Tensor : ",three_D_tensor)
print("three_D_tensor Dimension : ",three_D_tensor.ndim)
print("three_D_tensor Tensor Shape : ",three_D_tensor.shape)
```

```python
## GPU Accelration

if torch.cuda.is_available():
    gpu_tensor = vector.to('cuda')
    print("Tensor in GPU")
else:
    print("CUDA is not available")
```

```python
## Tensor Properties

print("Data Type of matrix : ",matrix.dtype)
print("Device of Tensor : ",matrix.device)
print("Size of matrix: ",matrix.size())
print("No. of elements in matrix: ",matrix.numel())
```

### INDEXING IN TENSORS

```python
# Accessing element in tensor
tensor=torch.tensor([[10,20,30],
                    [40,50 ,60],
                    [70,80,90]])
print(tensor.ndim)
```

```python
print(tensor[0,2])
print(tensor[1,1])
```

```python
tensor[0] ## First Row
```

```python
tensor[:,0] ## First Column
```

### SLICING IN TENSORS

```python
print("First 2 rows : ",tensor[:2])
```

```python
print("first 2 columns : \n",tensor[:,:2])
```

```python
print(tensor[1,1:3])
```

```python
## How to do fancy indexing (access 20 and 90)
## (0,1)-->20     (2,2)--->90
tensor[[0,2],[1,2]]

'''
0,2 --> rows
1,2 --> columns
 
'''
```

### Boolean Indexing

```python
mask = tensor>50
print(mask)

print(tensor[mask])
```

### Updation of values in tensor

```python
tensor[0,1] = 25
print(tensor)
```

```python
tensor[:,0] = torch.tensor([100,200,300])
print(tensor)
```

### ADVANCED SLICING

```python
indices = torch.tensor([0,2])
selected_rows = torch.index_select(tensor,dim=0,index=indices)
print(selected_rows)
```

### Using tensor with steps

```python
print(tensor[:,::2]) ## Columns on a step of 2
```

```python
'''
REVERSING THE TENSOR
'''
torch.flip(tensor,dims=(0,))
```

```python
torch.flip(tensor,dims=(1,))
```
