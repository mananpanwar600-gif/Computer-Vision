## Data Types
Floating used for most deep learning task

Integer used for indices or categorical data

Boolean used for mask or logical op

Comple Numbers used for Advanced Computation


```python
import torch
```

```python
default_tensor = torch.tensor([1.5,2.5,3.5])
print(default_tensor.dtype)

default_tensor_integer = torch.tensor([1,2,3])
print(default_tensor_integer.dtype)
```

```python
float_tensor = torch.tensor([1.5,2.5,3.5],dtype=torch.float64)
print(float_tensor.dtype)
```

```python
int_tensor = torch.tensor([1.5,2.5,3.5],dtype=torch.int32)
print(int_tensor.dtype)
print(int_tensor)
```

## Impact Of Data type on memory

```python
float32_tensor=torch.ones(1000,dtype=torch.float32)
float64_tensor=torch.ones(1000,dtype=torch.float64)

# tensor.element_size()
# element_size() --> Size of one elment in bytes
# nelement give sthe total number of elements in tensor

print("Memory used by float32: ",float32_tensor.element_size()*float32_tensor.nelement())
print("Memory used by float64: ",float64_tensor.element_size()*float64_tensor.nelement())
```

### Float 32 ---> Default for most NN 
### Float 64 ---> High Precison 
### Int 32 ---> General purpose
### Int 64 ---> tensor indices
### Float16  ---> Half percision of default floating dtype used to reduce mem usage
### Int 8 ---> Reduce mem usage
