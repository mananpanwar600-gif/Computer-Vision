### Noisy Image

```python
import matplotlib.pyplot as plt

## height and width -> 256*256

uniform_noise=torch.rand((256,256))
plt.imshow(uniform_noise,cmap='grey')
plt.title("Uniform Noise Image")
plt.axis('off')
plt.show()
```

```python
rgb_noise=torch.rand((256,256,3))
plt.imshow(rgb_noise)
plt.title("rgb_noise")
plt.axis('off')
plt.show()
```

```python
print("Shape:",rgb_noise.shape)
print("Data Type:",rgb_noise.dtype)
print("Max Value:",torch.max(rgb_noise))
```

## Tensors of zeros and ones

```python
torch.zeros(5)
```

```python
zero_tensor=torch.zeros(256,256)
plt.imshow(zero_tensor,cmap='grey',vmin=0,vmax=1)
plt.title("Dark Image")
plt.axis('off')
plt.show()

'''
0 -> Black
1 -> White
'''
```

```python
ones_tensor = torch.ones(256,256)
plt.imshow(ones_tensor,cmap='grey',vmin=0,vmax=1)
plt.title("White Image")
plt.axis('off')
plt.show()
```
